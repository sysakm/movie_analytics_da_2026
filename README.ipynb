# 🎬 Capstone Project — Netflix Multi-Table Analysis

**Your first realistic, multi-table data analysis project!**

Netflix wants deep insights into their catalog: what performs well, what viewers love, which actors and genres drive engagement, and what to invest in next.

You've been given **5 related CSV files** with information about titles, genres, cast, user ratings, and monthly viewership. Your job is to explore, clean, merge, analyze, and deliver a professional report.

> ⚠️ The data is realistic — expect missing values, duplicates, and denormalized structure. How you handle the multi-table joins is a **major** part of what's being evaluated.

## 📁 The Datasets

You have **5 CSV files** that relate to each other:

```
titles ──┬── genres        (one title → many genres)
         ├── cast          (one title → many actors)
         ├── ratings       (one title → many user ratings)
         └── viewership    (one title → many monthly view records)
```

### 📄 `titles.csv` — Main catalog

One row per title (movie or TV show).

| Column | Description |
|---|---|
| `title_id` | Unique identifier (primary key) |
| `type` | Movie or TV Show |
| `title` | Title name |
| `director` | Director (may be missing) |
| `country` | Country of production (may be missing) |
| `date_added` | Date added to catalog |
| `release_year` | Year released |
| `content_rating` | TV-MA, PG-13, etc. |
| `duration` | `"120 min"` for movies, `"3 Seasons"` for TV |

### 📄 `genres.csv` — Genre lookup

One row per (title, genre) combination. A title with 3 genres appears 3 times.

| Column | Description |
|---|---|
| `genre_id` | Unique row ID |
| `title_id` | Links to `titles` |
| `genre` | Genre name |

### 📄 `cast.csv` — Cast lookup

One row per (title, actor) combination.

| Column | Description |
|---|---|
| `cast_id` | Unique row ID |
| `title_id` | Links to `titles` |
| `actor_name` | Actor name (may be missing) |
| `role_type` | Lead or Supporting |

### 📄 `ratings.csv` — User ratings

One row per user rating. Many ratings per title.

| Column | Description |
|---|---|
| `rating_id` | Unique row ID |
| `title_id` | Links to `titles` |
| `user_id` | Anonymous user ID |
| `rating` | Integer 1–5 |
| `rating_date` | Date the rating was submitted |

### 📄 `viewership.csv` — Monthly views

One row per (title, month). Represents total views that month.

| Column | Description |
|---|---|
| `view_id` | Unique row ID |
| `title_id` | Links to `titles` |
| `month` | YYYY-MM |
| `views` | View count that month |

## ⚠️ CRITICAL Concept: "What Does One Row Represent?"

Each table has a **different grain**:

- `titles` → **one row = one title**
- `genres` → **one row = one (title, genre) pair**
- `cast` → **one row = one (title, actor) pair**
- `ratings` → **one row = one user's rating**
- `viewership` → **one row = one title's views in one month**

🚨 **Danger:** if you `merge` titles with ratings and then count titles, you'll count each title MANY times (once per rating). Always ask: *"After this merge, what does one row represent?"*

## 💡 Tips

- Explore **each table separately first** before merging anything.
- Convert `date_added` and `rating_date` to datetime early.
- For numeric duration: `df["duration_num"] = df["duration"].str.extract(r"(\d+)").astype(float)`.
- Use `merge()` with explicit `on="title_id"` and think about `how=` (inner/left).
- After merging, **verify row counts** — it's easy to accidentally 10x your data.
- For "most viewed" or "most rated", **aggregate first, then merge** to avoid duplicates.
- Every chart needs: **title, axis labels, readable ticks, and a written takeaway**.
- If you have to write the same analysis twice, wrap it in a function.
